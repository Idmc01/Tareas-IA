1. Random Sampling

```python
from sklearn.model_selection import train_test_split

cols = selected_features


X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.1765, random_state=42, stratify=y_temp)


X_train6 = X_train[cols].copy()
X_val6   = X_val[cols].copy()
X_test6  = X_test[cols].copy()
```

Manejo de outliers en las features seleccionadas

```python
p_low  = X_train6.quantile(0.05)
p_high = X_train6.quantile(0.95)

def winsorize(df, low, high):
    df = df.copy()
    for c in df.columns:
        df[c] = df[c].clip(low[c], high[c])
    return df

X_train_w = winsorize(X_train6, p_low, p_high)
X_val_w   = winsorize(X_val6,   p_low, p_high)
X_test_w  = winsorize(X_test6,  p_low, p_high)
```

```python
print("Train:\n", pd.Series(y_train).value_counts())
print("\nVal:\n",   pd.Series(y_val).value_counts())
print("\nTest:\n",  pd.Series(y_test).value_counts())
```

# C. Implementación de Regresión Logística

```python
!pip install -q torch
```

```python
# regresión logística binaria con PyTorch

import torch
import torch.nn as nn
import torch.optim as optim

# convertir los datos a tensores de PyTorch
X_train_t = torch.tensor(X_train_w.values, dtype=torch.float32)
y_train_t = torch.tensor(y_train.values, dtype=torch.float32).view(-1, 1)
X_val_t   = torch.tensor(X_val_w.values, dtype=torch.float32)
y_val_t   = torch.tensor(y_val.values, dtype=torch.float32).view(-1, 1)
X_test_t  = torch.tensor(X_test_w.values, dtype=torch.float32)
y_test_t  = torch.tensor(y_test.values, dtype=torch.float32).view(-1, 1)
```

```python
# definir el modelo
class LogisticRegression(nn.Module):
    def __init__(self, n_features):
        super().__init__()
        self.linear = nn.Linear(n_features, 1)
    def forward(self, x):
        return torch.sigmoid(self.linear(x))

model = LogisticRegression(X_train_t.shape[1])

# definir función de perdida y optimizador
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
# entrenamiento y toma de métricas

from sklearn.metrics import accuracy_score, precision_recall_fscore_support, roc_auc_score

def training(lr=0.01, n_epochs=3000):
    model = LogisticRegression(X_train_t.shape[1])
    criterion = nn.BCELoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)

    train_losses = []
    val_losses = []
    acc_val = None 

    for epoch in range(n_epochs):
        model.train()
        optimizer.zero_grad()
        outputs = model(X_train_t)
        loss = criterion(outputs, y_train_t)
        loss.backward()
        optimizer.step()
        train_losses.append(loss.item())

        model.eval()
        with torch.no_grad():
            val_pred = model(X_val_t)
            val_loss = criterion(val_pred, y_val_t).item()
            val_losses.append(val_loss)
            y_pred_label_val = (val_pred >= 0.5).float()
            acc_val = (y_pred_label_val == y_val_t).float().mean().item()

    with torch.no_grad():
        val_prob = val_pred.cpu().numpy().ravel()
        y_true  = y_val_t.cpu().numpy().ravel().astype(int)
        y_hat   = (val_prob >= 0.5).astype(int)

    acc  = accuracy_score(y_true, y_hat)
    prec, rec, f1, _ = precision_recall_fscore_support(
        y_true, y_hat, average='binary', zero_division=0
    )
    try:
        auc = roc_auc_score(y_true, val_prob)
    except ValueError:
        auc = float("nan")

    return {
        "lr": lr,
        "epochs": n_epochs,
        "val_loss": float(val_losses[-1]),
        "val_acc": float(acc),       
        "precision": float(prec),
        "recall": float(rec),
        "f1": float(f1),
        "auc": float(auc),
        "train_losses": train_losses,
        "val_losses": val_losses,
    }

# Define los 10 entrenamientos 
configs = [
    {"lr": 0.001, "n_epochs": 1500},
    {"lr": 0.001, "n_epochs": 3000},
    {"lr": 0.003, "n_epochs": 1500},
    {"lr": 0.003, "n_epochs": 3000},
    {"lr": 0.010, "n_epochs": 1000},
    {"lr": 0.010, "n_epochs": 2000},
    {"lr": 0.010, "n_epochs": 3000},
    {"lr": 0.030, "n_epochs": 1000},
    {"lr": 0.030, "n_epochs": 2000},
    {"lr": 0.030, "n_epochs": 3000},
]

results = []
for cfg in configs:
    r = training(lr=cfg["lr"], n_epochs=cfg["n_epochs"])
    results.append(r)

```

# Evaluación del modelo

### Diagramas de Training vs Validation de cada entrenamiento

```python
# Ajustes para dimensionar la figura según epochs
def sizeEpochs(epochs, minW=6, maxW=16, base=200):
    """
    Devuelve (width, height) para plt.figure. 
    width crece con epochs, acotado entre [minW, maxW].
    base controla la pendiente (más pequeño => crece más rápido).
    """
    width = min(max(minW, (epochs / base) * minW), maxW)
    return (width, 4.5)

for i, r in enumerate(results, start=1):
    tr = r['train_losses']
    vl = r['val_losses']
    epochs = len(tr)  
    w, h = sizeEpochs(epochs)

    plt.figure(figsize=(w, h))
    plt.plot(tr, label="Train Loss")
    plt.plot(vl, label="Validation Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Binary Cross Entropy Loss")
    plt.title(f"Run {i} — lr={r['lr']}, epochs={r['epochs']}")
    plt.legend()
    plt.grid(True, alpha=0.25)
    plt.tight_layout()
    plt.show()
```

# Análisis de resultados

### Seleccion del mejor entrenamiento

```python
sel_df = pd.DataFrame(results).copy()
sel_df['_auc'] = sel_df['auc'].fillna(-1)  
sel_df = sel_df.sort_values(
    by=['_auc', 'f1', 'val_acc', 'val_loss'],
    ascending=[False, False, False, True]
)

best_row = sel_df.iloc[0]
best_lr  = float(best_row['lr'])
best_ep  = int(best_row['epochs'])
best_idx = int(sel_df.index[0])  

print(f"Mejor corrida: Run {best_idx+1} | lr={best_lr} | epochs={best_ep}")
```

### Matriz de confusión

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

best_model = LogisticRegression(X_train_t.shape[1])
best_criterion = nn.BCELoss()
best_optim = optim.Adam(best_model.parameters(), lr=best_lr)

for i in range(best_ep):
    best_model.train()
    best_optim.zero_grad()
    out = best_model(X_train_t)
    loss = best_criterion(out, y_train_t)
    loss.backward()
    best_optim.step()

# Predicción en TEST
best_model.eval()
with torch.no_grad():
    test_pred = best_model(X_test_t)                 
    test_prob = test_pred.cpu().numpy().ravel()
    y_true    = y_test_t.cpu().numpy().ravel().astype(int)
    y_hat     = (test_prob >= 0.5).astype(int)       

# Métricas y matriz
cm  = confusion_matrix(y_true, y_hat, labels=[0,1])
acc = accuracy_score(y_true, y_hat)
pre = precision_score(y_true, y_hat, zero_division=0)
rec = recall_score(y_true, y_hat, zero_division=0)
f1  = f1_score(y_true, y_hat, zero_division=0)
auc = roc_auc_score(y_true, test_prob)

# Mostrar matriz de confusión
plt.figure(figsize=(5.5,4.5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Pred 0","Pred 1"], yticklabels=["True 0","True 1"])
plt.title(f"Confusion Matrix — best run (lr={best_lr}, epochs={best_ep})")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.tight_layout()
plt.show()

print(f"TEST metrics | acc={acc:.3f}  precision={pre:.3f}  recall={rec:.3f}  F1={f1:.3f}  AUC={auc:.3f}")
```
